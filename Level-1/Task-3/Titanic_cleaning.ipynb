**Outlier Detection**

```python
outlier_cols =df.select_dtypes(include = 'number').columns
```

```python
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    
    
```

```python
df.shape
```

Outlier Detection

The IQR method was used to identify potential outliers in the numerical variables Age, SibSp, Parch, and Fare. PassengerId, Survived, and Pclass were excluded because they are identifiers or categorical/ordinal variables rather than continuous measurements. The detected outliers were retained because they may represent legitimate passenger characteristics, such as higher fares or larger family groups. Removing them could result in the loss of meaningful information.

**Missing Values Handling**

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

Cabin: Missing values were replaced with "Unknown" because Cabin is a categorical identifier and its missing values cannot be meaningfully replaced using mean or median imputation. Keeping "Unknown" also preserves the information that the cabin was not recorded.

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

Missing Data Handling:


Age: Missing values were replaced with the median age because Age is a numerical variable and the median is less affected by extreme values.
Cabin: Missing values were replaced with "Unknown" because Cabin is a categorical variable with many missing values, and the missingness itself may contain useful information.
Embarked: Missing values were replaced with the mode because Embarked is a categorical variable and the most frequent category is an appropriate replacement.
Other columns: Columns without missing values were left unchanged.

**Before Cleaning vs After Cleaning**

```python
# ---------- BEFORE CLEANING ----------
before_rows = df.shape[0]
before_nulls = df.isnull().sum().sum()
before_duplicates = df.duplicated().sum()

# Expected logical data types
expected_dtypes = {
    'PassengerId': 'int',
    'Survived': 'int',
    'Pclass': 'int',
    'Name': 'string',
    'Sex': 'string',
    'Age': 'float',
    'SibSp': 'int',
    'Parch': 'int',
    'Ticket': 'string',
    'Fare': 'float',
    'Cabin': 'string',
    'Embarked': 'string'
}

# Check dtype accuracy before cleaning
before_dtype_accuracy = sum(
    str(df[col].dtype).lower().startswith(expected.lower())
    for col, expected in expected_dtypes.items()
) / len(expected_dtypes) * 100

# ---------- AFTER CLEANING ----------
after_rows = df.shape[0]
after_nulls = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()

# Check dtype accuracy after cleaning
after_dtype_accuracy = sum(
    str(df[col].dtype).lower().startswith(expected.lower())
    for col, expected in expected_dtypes.items()
) / len(expected_dtypes) * 100

# ---------- FINAL SUMMARY TABLE ----------
summary = pd.DataFrame({
    'Metric': [
        'Row Count',
        'Total Null Count',
        'Duplicate Count',
        'Dtype Accuracy'
    ],
    'Before Cleaning': [
        before_rows,
        before_nulls,
        before_duplicates,
        f'{before_dtype_accuracy:.2f}%'
    ],
    'After Cleaning': [
        after_rows,
        after_nulls,
        after_duplicates,
        f'{after_dtype_accuracy:.2f}%'
    ]
})

print(summary)
```
